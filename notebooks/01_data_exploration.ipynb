# 01 — Data Exploration
Explore all CIC-IDS2017 CSVs without modifying them.

```python
import pandas as pd, glob, os
RAW="../data/raw/MachineLearningCVE"
files=glob.glob(os.path.join(RAW,"*.csv"))
print("Number of CSV files:",len(files))
for f in files:
    df=pd.read_csv(f)
    df.columns=df.columns.str.strip()
    print(os.path.basename(f), df.shape)
    print(df["Label"].value_counts())
```
